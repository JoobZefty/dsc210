### Problem B2. Broadcasting (4 points)

The cell below pulls flipper lengths from the penguins dataset as a NumPy array.

````python
# FILL-IN  (Problem B2)
flippers_mm = penguins['flipper_length_mm'].dropna().to_numpy()

# (a) convert every measurement from millimeters to centimeters,
# using broadcasting
flippers_cm = flippers_mm.copy() / 10
print('first five (cm):', flippers_cm[:5])

# (b) how many flipper measurements are there?
print('count:', len(flippers_mm))

# (c) mean flipper length in cm, rounded to 2 decimal places
print('mean (cm):', round(sum(flippers_cm) / len(flippers_cm), 2))
````

### Problem B3. Boolean masks (4 points)

````python
# FILL-IN  (Problem B3)
mass_g = penguins['body_mass_g'].dropna().to_numpy()

# (a) a boolean mask for penguins over 4000 g
big = mass_g > 4000
print('first five of the mask:', big[:5])

# (b) how many penguins are over 4000 g?
print('count over 4000 g:', sum(big))

# (c) the fraction of all penguins over 4000 g
#     (recall: the mean of a boolean array is a proportion)
print('fraction:', round(big.mean(), 3))

# (d) the mean mass of ONLY the penguins over 4000 g
print('mean of the big ones:', round(mass_g[big].mean(), 1))
````

Parts (c) and (d) both use the mask, but they answer different questions. State each question in plain English.

**Your answer:** (c) What proportion of penguins are big (d) What is the average mass of big penguins.

(a) (3 points) Write code to compute: the number of rainy days (rainfall greater than 0), the fraction of days that were rainy, the mean rainfall over **all** days, and the mean rainfall over **rainy days only**. Round the means to 2 decimal places.

````python
# Problem D2(a): your code here
rain = np.array([0, 0, 3.2, 11.4, 0, 0.6, 0,
                 0, 0, 22.1, 4.5, 0, 0, 1.1,
                 0, 8.3, 0, 0, 2.7, 0, 0])
#(a) (3 points) Write code to compute: the number of rainy days (rainfall greater than 0), the fraction of days that were rainy, the mean rainfall over all days, and the mean rainfall over rainy days only.
#Round the means to 2 decimal places.
print('Number of rainy days:', sum(rain > 0))
print('Fraction of rainy days:',round(sum(rain > 0) / len(rain), 2))
print('Mean rainfall over all days:', round(sum(rain) / len(rain), 2))
print('Mean rainfall over rainy days:', round(sum(rain[rain > 0]) / sum(rain > 0), 2))
````
