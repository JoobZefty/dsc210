A tree considers two candidate questions: **Split 1: `bang` < 0.2** and **Split 2: `bang` < 0.4**.

Check your arithmetic with the cell below.

**Your answer:**

(a) G = 1 - (0.6^2 + 0.4^2) = 0.48

(b)

**Split 0.2**: `left child:` 4n / 0y / Gini = 0, `right child:` 2n / 4y / Gini = 0.444, `Weighted Child Gini:` $\frac{n_L}{8} G_L + \frac{n_R}{8} G_R$ = (0.4 * 0)+(0.6 * 0.444) = 0.267, `Gini Gain:` Parent - weighed child = 0.48 - 0.267 = 0.213

**Split 0.4**: `left child:` 5n / 0y / Gini = 0, `right child:` 1n / 4y / Gini = 0.32, `Weighted Child Gini:` $\frac{n_L}{8} G_L + \frac{n_R}{8} G_R$ = (0.5 * 0)+(0.5 * 0.32) = 0.16, `Gini Gain:` Parent - weighed child = 0.48 - 0.16 = 0.32

The tree will pick the second split.

(c) The second one has a larger gain because the other child has a high proportion of similar values. This is because the inclusion/exclusion of email number 2734 (bang = 0.027,  yesno = n)

(d) The tree could filter next on a different feature of the email in which email number 3553 is sectioned from the rest.

````python
# FILL-IN  (Problem B1): check your arithmetic
def gini(n_spam, n_not):
    n = n_spam + n_not
    if n == 0:
        return 0.0
    return 1 - ((n_spam / n) ** 2 + (n_not / n) ** 2)

parent = gini(4, 6)
print('parent Gini:', round(parent, 4))

# Split 1: bang < 0.2.  Fill in (number of spam, number not spam) for each child.
left1, right1 = gini(0, 4), gini(4, 2)
weighted1 = 4/10 * left1 + 6/10 * right1
print('split 1: left', round(left1, 4), ' right', round(right1, 4), ' weighted', round(weighted1, 4), ' gain', round(parent - weighted1, 4))

# Split 2: bang < 0.4
left2, right2 = gini(0, 5), gini(4, 1)
weighted2 = 5/10 * left2 + 5/10 * right2
print('split 2: left', round(left2, 4), ' right', round(right2, 4), ' weighted', round(weighted2, 4), ' gain', round(parent - weighted2, 4))
````
